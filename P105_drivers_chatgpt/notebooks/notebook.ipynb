# Drivers con asistencia de ChatGPT

Análisis de `drivers.csv` y `timesheet.csv` construido a partir de prompts a ChatGPT.

**Prompt:** Carga los archivos drivers.csv y timesheet.csv con pandas y muestra las primeras filas.

```python
import matplotlib.pyplot as plt
import pandas as pd

drivers = pd.read_csv("../data/drivers.csv")
timesheet = pd.read_csv("../data/timesheet.csv")
display(drivers.head(), timesheet.head())
```

**Prompt:** Calcula el total de horas y millas por conductor y agrega su nombre.

```python
summary = (
    timesheet.groupby("driverId", as_index=False)[["hours-logged", "miles-logged"]]
    .sum()
    .merge(drivers[["driverId", "name"]], on="driverId")
)
summary
```

**Prompt:** Calcula las millas por hora de cada conductor y ordénalos de mayor a menor.

```python
summary["miles-per-hour"] = summary["miles-logged"] / summary["hours-logged"]
summary.sort_values("miles-per-hour", ascending=False).head(10)
```

**Prompt:** Grafica los 10 conductores con más millas en barras horizontales.

```python
top10 = summary.nlargest(10, "miles-logged").set_index("name")

top10["miles-logged"].plot.barh(color="tab:blue", alpha=0.6)
plt.gca().invert_yaxis()
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)
plt.xlabel("miles-logged")
plt.show()
```
